```python
import pandas as pd
import numpy as np

df = pd.read_csv("teaching_prices_messy.csv")
original_df = df.copy() 

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData types and non-null counts:")
df.info()

print("\nMissing values:")
display(df.isna().sum().to_frame("missing_count"))

print("\nExact duplicate rows:", df.duplicated().sum())

print("\nDescriptive summary:")
display(df.describe(include="all"))
```

```python
for column in df.select_dtypes(include=["object", "string"]).columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False))
```

```python
if "price" in df.columns:
    display(df["price"].value_counts(dropna=False).to_frame("frequency"))
    print("\nEntries containing commas:")
    display(df[df["price"].astype(str).str.contains(",", na=False)])
```

```python
if "price" in df.columns:
    price_text = df["price"].astype("string").str.replace(",", "", regex=False)
    price_check = pd.to_numeric(price_text, errors="coerce")
    print(price_check.describe())
    print("\nLargest values:")
    display(price_check.nlargest(10))
    print("\nMissing/unconvertible price rows:")
    display(df[price_check.isna()])
```

```python
display(df[df.duplicated(keep=False)])
```

```python
clean_df = df.copy()
display(clean_df.head())
```

```python
print("Original rows:", len(original_df))
print("Current cleaned rows:", len(clean_df))
print("Duplicates remaining:", clean_df.duplicated().sum())
```

```python
print("ORIGINAL DATA")
print("Shape:", original_df.shape)
print("Missing values:")
display(original_df.isna().sum())
print("Duplicate rows:", original_df.duplicated().sum())

print("\nCURRENT CLEANED DATA")
print("Shape:", clean_df.shape)
print("Missing values:")
display(clean_df.isna().sum())
print("Duplicate rows:", clean_df.duplicated().sum())
print("\nPreview:")
display(clean_df.head(10))
print("\nData types:")
clean_df.info()
```

```python
clean_df.to_csv("cleaned_teaching_prices.csv", index=False)
print("Saved cleaned_teaching_prices.csv")
```
